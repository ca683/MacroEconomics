```julia
%%shell
set -e

#---------------------------------------------------#
JULIA_VERSION="1.6.0" # any version ≥ 0.7.0
JULIA_PACKAGES="IJulia Plots"
JULIA_PACKAGES_IF_GPU="CUDA" # or CuArrays for older Julia versions
JULIA_NUM_THREADS=2
#---------------------------------------------------#

if [ -n "$COLAB_GPU" ] && [ -z `which julia` ]; then
  # Install Julia
  JULIA_VER=`cut -d '.' -f -2 <<< "$JULIA_VERSION"`
  echo "Installing Julia $JULIA_VERSION on the current Colab Runtime..."
  BASE_URL="https://julialang-s3.julialang.org/bin/linux/x64"
  URL="$BASE_URL/$JULIA_VER/julia-$JULIA_VERSION-linux-x86_64.tar.gz"
  wget -nv $URL -O /tmp/julia.tar.gz # -nv means "not verbose"
  tar -x -f /tmp/julia.tar.gz -C /usr/local --strip-components 1
  rm /tmp/julia.tar.gz

  # Install Packages
  if [ "$COLAB_GPU" = "1" ]; then
      JULIA_PACKAGES="$JULIA_PACKAGES $JULIA_PACKAGES_IF_GPU"
  fi
  for PKG in `echo $JULIA_PACKAGES`; do
    echo "Installing Julia package $PKG..."
    julia -e 'using Pkg; pkg"add '$PKG'; precompile;"' &> /dev/null
  done

  # Install kernel and rename it to "julia"
  echo "Installing IJulia kernel..."
  julia -e 'using IJulia; IJulia.installkernel("julia", env=Dict(
      "JULIA_NUM_THREADS"=>"'"$JULIA_NUM_THREADS"'"))'
  KERNEL_DIR=`julia -e "using IJulia; print(IJulia.kerneldir())"`
  KERNEL_NAME=`ls -d "$KERNEL_DIR"/julia*`
  mv -f $KERNEL_NAME "$KERNEL_DIR"/julia  

  echo ''
  echo "Success! Please reload this page and jump to the next section."
fi
```

```julia
# Calling all packages at the beginning is good practice
using Plots # Notice first time takes long (20 secs approx)
```

```julia
# Let's create some data to plot
x = 1:10
y = x.^3
y1 = x.^2
```

```julia
# One way of plotting with labels
plot(x, y, label="x^3", linewidth = 3, legend = :topleft) # To create a LaTeX string, we have to write L before the string and put the LaTeX code inside $
plot!(x, y1, label="x^2", linewidth = 3)
xlabel!("Number of chocolate bars")
ylabel!("Weight")
title!("Obesity Figure 1")
```

```julia
# A second way of plotting
plot(x, [y, y1],
   label = ["x^3" "x^2"],
   linewidth = 3,
   legend = :topleft, # or no legend, legend = false
   xlabel = "Number of chocolate bars",
   ylabel = "Weight",
   title = "Obesity Figure 2")
```

```julia
# A third way of plotting that uses nice LaTeX font. If you haven't done so, download the LaTeXStrings package.
using LaTeXStrings
plot(x, [y, y1], label = [L"$x^3$" L"$x^2$"],
   linewidth = 3,
   legend = :topleft,
   xlabel = L"\textrm{Number of chocolate bars}",
   ylabel = L"\textrm{Weight}",
   title = L"\textrm{Obesity Figure 3 (using nice LaTeX font)}")
```

```julia
# For Loops
# Example: Take for example depreciation, we could just type out the sequence of calculations above by hand,
# one after the other.
α = 0.143;  # Rate of depreciation (the greek letter α is a Unicode character. You get it by typing \alpha and clicking the TAB button)
m = 4;  # Number of rows
n = 1;  # Number of columns
U1 = zeros(m,n); # Prepopulate a matrix where we will store our results
U[1] = 0.5;
U[2] = (1-α)*U[1];
U[3] = (1-α)*U[2];
U[4] = (1-α)*U[3];
```

```julia
# The for loop provides a much more efficient way of doing this calculation:
α = 0.143;  # Rate of depreciation
m = 4;  # Number of rows
n = 1;  # Number of columns
U = zeros(m,n); # Prepopulate a matrix where we will store our results
U[1] = 0.5;
for k = 2:4
   U[k] = (1-α)*U[k-1]; # Notice it doesn't work if it is not indented
end
```

```julia
# An equivalent way of writing the loop above
α = 0.143;  # Rate of depreciation
m = 4;  # Number of rows
n = 1;  # Number of columns
U = zeros(m,n); # Prepopulate a matrix where we will store our results
U[1] = 0.5;
for k=1:3
   U[k+1] = (1-α)*U[k];
end
```

```julia
# For loops are not restricted to a single operation
x = 4:8;
m = 5;  # Number of rows
n = 1;  # Number of columns
s = zeros(m,n); # Prepopulate a matrix where we will store our results
c = zeros(m,n); # Prepopulate a matrix where we will store our results
f = zeros(m,n); # Prepopulate a matrix where we will store our results
for k=1:m
   s[k] = x[k]^2;
   c[k] = x[k]^3;
   f[k] = x[k]^4;
end
```

```julia
# Now a longer For Loop. Notice this would be a pain by hand.
α = 0.143;
t = 0:25;
m = 26;  # Number of rows
n = 1;  # Number of columns
r = zeros(m,n); # Prepopulate a matrix where we will store our results
r[1] = 1;
for k = 2:m;
   r[k] = (1-α)*r[k-1];
end
plot(t,r)
xlabel!("Time period")
ylabel!("U(t)/U(0)")
```

```julia
# Notice we can nest loops
α = 0.143;
t = 0:25;
m = 26;  # Number of rows
n = 5;  # Number of columns
r = zeros(m,n); # Prepopulate a matrix where we will store our results
r[1,:] .= 1;
for k = 2 : m
   # @show k # @show command allows you to see output in each iteration
   for j = 2 : n
      r[k,j] = (1-α)*r[k-1,j-1]*j
   end
end
plot(t,r)
```
