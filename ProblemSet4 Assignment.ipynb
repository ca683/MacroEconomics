## Q1A: STEADY STATE CONSUMPTION $C*$ AND CAPITAL $k*$

CALCULATING THE STEADY-STATE CONSUMPTION $C^*$


Given the capital accumulation equation

$$
k_{t+1}=f(k_t)+(1-\delta)k_t-c_t
\qquad \text{Eq. (28)}
$$

from eq. 2, steady state capital and consumption remain constant. Substituting this into Eq. (29):

$$
k^*=f(k^*)+(1-\delta)k^*-c^*
\qquad \text{Eq. (30)}
$$


$$
k^*=f(k^*)+k^*-\delta k^*-c^*
$$


$$
0=f(k^*)-\delta k^*-c^*
\qquad \text{Eq. (31)}
$$


$$
c^*=f(k^*)-\delta k^*
\qquad \text{Eq. (32)}
$$


Given the production function $f(k)=k^{0.3}$ and depreciation rate $\delta=0.05$:

$$
c^*=(k^*)^{0.3}-0.05k^*
\qquad \text{Eq. (33)}
$$

Substituting the steady-state capital $k^*=4.803987$:

$$
c^*=(4.803987)^{0.3}-0.05(4.803987)
$$

$$
c^*\approx1.601329-0.240199
$$

Therefore, the steady-state consumption becomes;

$$
c^*\approx1.361130
\qquad \text{Eq. (34)}
$$

Capital $k^*$  = 4.803987



Consumption $C^*$  = 1.361130

### Translating the above calculation into Julia

```julia
# Define variables
A = 1.0
alpha = 0.3
sigma = 2.0
rho = 0.05
delta = 0.05

# Compute the discount factor using the rate of time preference
beta = 1 / (1 + rho)

# Using the formulae from Euler eqution to compute the Steady-State Capital 
k_star = (alpha * A / (rho + delta))^(1 / (1 - alpha))

# Calculate the Steady-State Consumption 
c_star = A * k_star^alpha - delta * k_star

# Results for Steady-State Consumption and Capital
println("Steady-state capital = ", round(k_star, digits=6))
println("Steady-state consumption = ", round(c_star, digits=6))
```

# Q1B: PLOT FOR SADDLE PATH

### Calculation for Initial Capital

```julia
# Define values given inn the question and use rho to compute for beta
A = 1.0
alpha = 0.3
sigma = 2.0
rho = 0.05
delta = 0.05
beta = 1 / (1 + rho)

# Formula to calculate the steady-state Capital and Consumption equation
k_star = (alpha * A / (rho + delta))^(1 / (1 - alpha))
c_star = A * k_star^alpha - delta * k_star

# Define the Initial capital to half of steady-state capital,  0.5 per the problem set specification
k0 = 0.5 * k_star

# Output for the Initial capital result
println("Initial capital = ", round(k0, digits=6))
```

### Establishing the Initial Consumption

```julia
# Define the number of periods.
T = 60

# Compute k0 as specified in 1b
k0 = 0.5 * k_star

# Initial capital rounded to 6 decimal places
println("Initial capital = ", round(k0, digits=6))

# Create space to hold cpapital and consumption
k = zeros(T + 1)
c = zeros(T + 1)

# Assign initial capital 
k[1] = k0

# Display the initial capital and consumption status 
println("Capital at t = 0: ", k[1])  
println("Consumption at t = 0: Not yet determined")
```

### Obtaining the sequence for {$c_t, k_t$}

```julia
# Display the first five periods of the simulation.
println("Period     Capital       Consumption") # label the columns for the period, capital and consumption

for t in 1:5     # loop through the first five capital and consumption values

    println(
        rpad(t - 1, 11),  # adjust the period number starting from zero
        rpad(round(k[t], digits=6), 14), # round the capital values to 6 decimal places
        round(c[t], digits=6) # print consumption rounded to 6 decimal places
    )

end
```

### Capital and Consumption Simulation

```julia
# Select trial value for initial consumption.
c0 = 1.0

# Assign initial values of capital and consumption
k[1] = k0
c[1] = c0

# Use a loop to simulate the calculations for 60 periods.

for t in 1:T

    # calculate capital using the capital accumulation equation

    k[t+1] = A * k[t]^alpha + (1 - delta) * k[t] - c[t]

    # stop the loop if capital is not positive

    if k[t+1] <= 0
        println("Capital became non-positive at period ", t)
        break
    end

    # calculate consumption in the Euler equation

    c[t+1] = c[t] * (
        beta * (alpha * A * k[t+1]^(alpha - 1) + 1 - delta)
    )^(1 / sigma)

end

# STEP 7: Display the first five observations of the simulated path

println("Period     Capital       Consumption")

for t in 1:5
    if k[t] > 0 && c[t] > 0
        println(
            rpad(t - 1, 11),
            rpad(round(k[t], digits=6), 14),
            round(c[t], digits=6)
        )
    end
end
```

### Adopting Bisection for the Shooting Method 

```julia
# Compute for the derivative needed for the stable path 
g_prime = alpha * A * k_star^(alpha - 1) + 1 - delta

h_prime = (beta / sigma) * alpha * A *
          (alpha - 1) * k_star^(alpha - 2)

# Construct the Jacobian matrix for analysing the steady state
J = [g_prime              -1.0;
     c_star*h_prime*g_prime  1 - c_star*h_prime]

# Use the stable eigenvectors and compute the saddle-path slope
using LinearAlgebra

E = eigen(J)

# The stable eigenvalue is the one smaller than 1.
stable_index = findfirst(x -> abs(x) < 1, E.values)

slope = E.vectors[2, stable_index] /
        E.vectors[1, stable_index]


 
# Define the lower and upper limits for inital consumption
c_low = 0.0
c_high = A * k0^alpha + (1 - delta) * k0


# Apply 60 bisection iterations to refine the initial consumption estimate
for guess in 1:60
    c0_trial = (c_low + c_high) / 2

    # Set arrays for both capital and consumption.
    k_trial = zeros(T + 1)
    c_trial = zeros(T + 1)

    # Create the initial values for the arrays.
    k_trial[1] = k0
    c_trial[1] = c0_trial

    # Check if the trial has been classified.
    classified = false

    # Simulate the economy and stop if capital becomes non-positive.
    for t in 1:T

        # Compute for the next-period capital.
        k_trial[t+1] = A * k_trial[t]^alpha +
                       (1 - delta) * k_trial[t] - c_trial[t]

        # Define Non-positive capital to be high consumption
        if k_trial[t+1] <= 0
            c_high = c0_trial
            classified = true
            break
        end

        # Calculate next-period consumption using the Euler equation.
        c_trial[t+1] = c_trial[t] * (
            beta * (alpha * A * k_trial[t+1]^(alpha - 1) +
            1 - delta)
        )^(1 / sigma)

        # Classify the trial path using the steady-state crossing conditions.
        if c_trial[t+1] >= c_star && k_trial[t+1] < k_star
            c_high = c0_trial
            classified = true
            break
        end

        # When the capital crosses above steady state first it implies initial capital is too low
        if k_trial[t+1] >= k_star && c_trial[t+1] < c_star
            c_low = c0_trial
            classified = true
            break
        end

    end

    # Make comparism with the final vlaues and the stable-path line
    if !classified

        # The stable-path line accounts for both k and c.
        target_c = c_star + slope * (k_trial[end] - k_star)

        if c_trial[end] > target_c
            c_high = c0_trial
        else
            c_low = c0_trial
        end

    end

end


# Report the estimated saddle-path consumption and final bisection bounds.
c0_saddle = (c_low + c_high) / 2


# Print the results.
println("SHOOTING METHOD RESULTS")
println("Estimated initial consumption = ",
        round(c0_saddle, digits=8))
println("Lower bound = ", c_low)
println("Upper bound = ", c_high)
println("Interval width = ", c_high - c_low)
```

### Authenticating The Estimated Initial Consumption

```julia
# Initialise fresh arrays for the 60-period simulation.
k_check = zeros(T + 1)
c_check = zeros(T + 1)

# Set the initial conditions for the verification path.
k_check[1] = k0
c_check[1] = c0_saddle

# Simulate the economy using the estimated 60 periods.
for t in 1:T

    # Compute for the next period's capital.
    k_check[t+1] = A * k_check[t]^alpha +
                   (1 - delta) * k_check[t] - c_check[t]

    # Stop if capital becomes non-positive.
    if k_check[t+1] <= 0
        println("Capital became non-positive at period ", t)
        break
    end

    # Update the consumption using the Euler equation
    c_check[t+1] = c_check[t] * (
        beta * (alpha * A * k_check[t+1]^(alpha - 1) +
        1 - delta)
    )^(1 / sigma)

end

# Compare the final results of capital and consumption with the steady state.
println("Estimated initial consumption = ", c0_saddle)
println("Final capital = ", k_check[end])
println("Steady-state capital = ", k_star)
println("Final consumption = ", c_check[end])
println("Steady-state consumption = ", c_star)
```

# 1(D): CONSTRUCTION OF THE POLICY FUNCTION

### Creation of a Grid for the Initial Capital Values

```julia
#Generate 20 capital values between 50% and 150% of a state capital.
k_grid_policy = range(0.5 * k_star,
                      1.5 * k_star,
                      length=20)

# label the heading of the capital values
println("Capital grid for the policy function:")

# print each capital values 
for k_value in k_grid_policy
    println(round(k_value, digits=6))
end
```

### Storing the Optimal Consumption Values 

```julia
# create array to store the consumption values
c_policy = zeros(length(k_grid_policy))

# check the number of capital and consumption values.
println("Number of capital values = ", length(k_grid_policy))
println("Number of consumption values = ", length(c_policy))
```

### Calculation of Policy Function

```julia
# Apply the shooting method for each capital value.
for i in eachindex(k_grid_policy)

    # Select the initial capital 
    k_initial = k_grid_policy[i]

    # Use the steady-state solution when capital equals its equilibrium values. 
    if abs(k_initial - k_star) < 1e-10
        c_policy[i] = c_star
        continue
    end

    # Set the lower and upper consumption bounds. 
    c_low_policy = 0.0
    c_high_policy = A * k_initial^alpha +
                    (1 - delta) * k_initial

    # Use bisection to estimate initial consumption.
    for guess in 1:60

        c_trial_0 = (c_low_policy + c_high_policy) / 2

        # set arrays for the trial simulation.
        k_trial = zeros(T + 1)
        c_trial = zeros(T + 1)

        # create initial capital and consumption.
        k_trial[1] = k_initial
        c_trial[1] = c_trial_0

        classified = false

        # Simulate the economy.
        for t in 1:T

            # Calculate next-period capital.
            k_trial[t+1] = A * k_trial[t]^alpha +
                           (1 - delta) * k_trial[t] - c_trial[t]

            # check whether the Consumption estimate is too high or low
            if k_trial[t+1] <= 0
                c_high_policy = c_trial_0
                classified = true
                break
            end

            # Calculate next-period consumption.
            c_trial[t+1] = c_trial[t] * (
                beta * (alpha * A * k_trial[t+1]^(alpha - 1) +
                1 - delta)
            )^(1 / sigma)

            # Consumption is too high if it crosses c_star
            # while capital remains below k_star.
            if c_trial[t+1] >= c_star && k_trial[t+1] < k_star
                c_high_policy = c_trial_0
                classified = true
                break
            end

            # Consumption is too low if capital crosses k_star while consumption remains below c_star.
            if k_trial[t+1] >= k_star && c_trial[t+1] < c_star
                c_low_policy = c_trial_0
                classified = true
                break
            end

            # Above steady state, reverse the crossing tests.
            if k_trial[t+1] <= k_star && c_trial[t+1] > c_star
                c_high_policy = c_trial_0
                classified = true
                break
            end

            if c_trial[t+1] <= c_star && k_trial[t+1] > k_star
                c_low_policy = c_trial_0
                classified = true
                break
            end

        end

        # Use the local stable-path approximation if no crossing occurs.
        if !classified

            target_c = c_star + slope * (k_trial[end] - k_star)

            if c_trial[end] > target_c
                c_high_policy = c_trial_0
            else
                c_low_policy = c_trial_0
            end

        end

    end

    # Store the estimated consumption for this capital value
    c_policy[i] = (c_low_policy + c_high_policy) / 2

end

# Display the capital and estimated consumption vlaues
println(rpad("Capital", 16), "Optimal consumption")

for i in eachindex(k_grid_policy)
    println(rpad(round(k_grid_policy[i], digits=6), 16),
            round(c_policy[i], digits=6))
end
```
